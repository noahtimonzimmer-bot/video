```
# speed test: 30 frames with the GPU (angle-egl). If this fails or is slower than ~2 s/frame, the next cell falls back.
import subprocess, time
t=time.time()
r=subprocess.run('npx remotion render src/index.ts ShirtShort out/test.mp4 --frames=600-629 --gl=angle-egl --concurrency=2', shell=True)
GL = 'angle-egl' if r.returncode==0 else 'swangle'
print('GL mode:', GL, ' time for 30 frames:', round(time.time()-t), 's')
```

```
# full render, 1080x1920, with sound
!npx remotion render src/index.ts ShirtShort out/ShirtGuy_1080.mp4 --codec h264 --crf 16 --gl={GL} --concurrency=2
```
